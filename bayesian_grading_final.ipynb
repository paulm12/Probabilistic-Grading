### For True-False Questions

```python
def calc_score_quadratic_tf(s, k):
    """
    Calculates the score for a true-false question given the students' response s and the true answer k
    s: Students' response (The expected score is maximized when this equals the students' creeence)
    k: A boolean value for the answer key (k=1 if the statement is true and k=0 if the statement is false)
    """
    return (1 - 2*(s - k)**2)

print(calc_score_quadratic_tf(0, 0))
print(calc_score_quadratic_tf(1, 1))

print(calc_score_quadratic_tf(0.9, 1))
print(calc_score_quadratic_tf(0.7, 1))
print(calc_score_quadratic_tf(0.5, 1))
print(calc_score_quadratic_tf(0.5, 0))
print(calc_score_quadratic_tf(0.3, 1))


print(calc_score_quadratic_tf(0, 1))
print(calc_score_quadratic_tf(1, 0))
```

```python
def calc_expected_tf_score(s, c):
    """
    Calculates the expected score of a student's response for true-false questios
    s -- What the student reports 
    c -- The students' true creedence about whether the answer is true or false
    Notice how the score is maximized when s = c; this is because calc_score_quadratic_tf is a proper scoring rule
    """
    score = c * calc_score_quadratic_tf(s=s, k=1) + (1 - c) * calc_score_quadratic_tf(s=s, k=0)
    return score


dist = calc_expected_tf_score(1, 1)
print(dist)

dist = calc_expected_tf_score(0.6, 0.6)
print(dist)

dist = calc_expected_tf_score(1, 0.6)
print(dist)

dist = calc_expected_tf_score(0.5, 0.5)
print(dist)

dist = calc_expected_tf_score(1, 0.5)
print(dist)

dist = calc_expected_tf_score(1, 0.6)
print(dist)

dist = calc_expected_tf_score(0, .4)
print(dist)

dist = calc_expected_tf_score(.4, .4)
print(dist)

dist = calc_expected_tf_score(1, 0)
print(dist)

dist = calc_expected_tf_score(0, 1)
print(dist)
```

### Calculate Students' Expected Score:

```python
def calc_expected_mc_score(s, c):
    """
    Calculates the expected score of a student's response for multiple choice questions
    s -- A vector of the students' reported values for each answer
    c -- The students' true creedences about whether each selection is the correct one 
    Notice how the score is maximized when s = c; this is because calc_score_quadratic_mc is a proper scoring rule"
    """
    res = []
    for i in range(len(c)):
        res.append(calc_score_quadratic_mc(s, i) * c[i])
    return res

overconfident_vector = [1, 0, 0, 0]
overconfident_wrong = [0, 1, 0, 0]

alice_res_honest = calc_expected_mc_score(alices_creedences, alices_creedences)
alices_res_overconfident = calc_expected_mc_score(overconfident_vector, alices_creedences)
print(f"Expected score if Alice is Honest {np.sum(alice_res_honest)}")
print(f"Expected score if Alice is Overconfident {np.sum(alices_res_overconfident)}")

bob_res_honest = calc_expected_mc_score(bob, bob)
bob_res_overconfident = calc_expected_mc_score(overconfident_vector, bob)
print(f"Expected score if Bob is Honest {np.sum(bob_res_honest)}")
print(f"Expected score if Bob is Overconfident {np.sum(bob_res_overconfident)}")

charles_res_honest = calc_expected_mc_score(charles, charles)
charles_res_overconfident = calc_expected_mc_score(overconfident_vector, charles)
print(f"Expected score if Charles is Honest {np.sum(charles_res_honest)}")
print(f"Expected score if Charles is Overconfident {np.sum(charles_res_overconfident)}")

dan_res_honest = calc_expected_mc_score(dans_creedences, dans_creedences)
dan_res_overconfident = calc_expected_mc_score(overconfident_wrong, dans_creedences)

print(f"Expected score if Dan is Honest {np.sum(dan_res_honest)}")
print(f"Expected score if Dan is Overconfident {np.sum(dan_res_overconfident)}")



```
